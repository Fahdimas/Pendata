# Business Understanding

Tahap ini merupakan fase pertama dalam metode CRISP-DM. Fokus pada tahap ini adalah memahami tujuan dan kebutuhan dari perspektif bisnis atau penelitian, kemudian menerjemahkannya ke dalam definisi masalah data mining.

## 1. Menentukan Tujuan Bisnis (Business Objectives)
Tujuan utama dari analisis dataset Iris ini adalah untuk membangun model klasifikasi yang mampu membedakan jenis bunga Iris secara akurat berdasarkan karakteristik fisiknya. Hal ini penting dalam bidang botani untuk identifikasi spesies secara otomatis.

## 2. Menilai Situasi (Assess Situation)
Dataset Iris yang digunakan adalah dataset standar yang sudah teruji kualitasnya. Tantangan dalam situasi ini adalah adanya kemiripan fitur antara spesies *Iris-versicolor* dan *Iris-virginica*, sehingga diperlukan metode analisis yang tepat untuk membedakannya.

## 3. Menentukan Tujuan Data Mining (Data Mining Goals)
Tujuan teknis dari proyek ini adalah:
* Melakukan eksplorasi data untuk melihat pola penyebaran spesies.
* Mengidentifikasi fitur (seperti panjang/lebar mahkota) yang paling berpengaruh terhadap jenis spesies.
* Membangun model yang memiliki tingkat akurasi tinggi dalam klasifikasi.

## 4. Membuat Rencana Proyek (Project Plan)
Langkah-langkah yang akan dilakukan dalam proyek ini meliputi:
1. **Data Understanding**: Mengenal karakteristik data (Orange & Python).
2. **Data Preparation**: Membersihkan atau memformat data jika diperlukan.
3. **Modeling**: Menerapkan algoritma klasifikasi.
4. **Evaluation**: Menghitung akurasi model.
