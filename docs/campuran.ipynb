# Mengukur Jarak Tipe Data Campuran (Gower Distance)

Untuk mendemonstrasikan perhitungan jarak pada tipe data campuran secara manual, kita menggunakan subset dari dataset publik. Dataset yang digunakan adalah **Telco Customer Churn** yang dapat diakses melalui tautan berikut: 
🔗 **[Kaggle Dataset: Telco Customer Churn](https://www.kaggle.com/datasets/blastchar/telco-customer-churn)**.

Kita mengambil tepat 4 kolom yang mewakili 4 tipe data yang berbeda agar perhitungannya dapat diurai secara spesifik.

## 1. Subset Dataset Telco Churn (4 Tipe Data)
Berikut adalah 3 sampel data pelanggan yang akan kita hitung jaraknya:

| CustomerID | MonthlyCharges (Numerik) | Churn (Biner) | PaymentMethod (Kategori) | Contract (Ordinal) |
|---|---|---|---|---|
| **C1** | 20.0 | No | Mailed check | Month-to-month (1) |
| **C2** | 70.0 | Yes | Electronic check | Two year (3) |
| **C3** | 25.0 | No | Mailed check | One year (2) |

**Informasi Rentang (Range) Keseluruhan Data:**
* **MonthlyCharges (Numerik):** Asumsi tagihan maksimal 120 dan minimal 20, maka $Range = 100$.
* **Contract (Ordinal):** Kontrak terlama adalah tingkat 3, terpendek tingkat 1, maka $Range = 3 - 1 = 2$.

---

## 2. Perhitungan Jarak Manual: Pelanggan 1 (C1) ke Pelanggan 2 (C2)

Berdasarkan formula Gower, kita menghitung jarak (ketidakmiripan) antar fitur satu per satu berdasarkan tipe datanya, kemudian menjumlahkannya.

### **A. Jarak Fitur 1: MonthlyCharges (Numerik)**
**Rumus:**
$$d_{ij}^{(f)} = \frac{|x_{if} - x_{jf}|}{R_f}$$

**Keterangan:**
* $x_{if}$: Nilai fitur pelanggan pertama (C1).
* $x_{jf}$: Nilai fitur pelanggan kedua (C2).
* $R_f$: Rentang (range) maksimal dari fitur tersebut.

**Perhitungan:**
* Nilai C1 = 20.0, C2 = 70.0, $R_f = 100$
* $d_1 = \frac{|20.0 - 70.0|}{100} = \frac{50}{100} = 0.5$

### **B. Jarak Fitur 2: Churn (Biner)**
**Rumus:**
$$d_{ij}^{(f)} = \begin{cases} 0, & \text{jika } x_{if} = x_{jf} \\ 1, & \text{jika } x_{if} \neq x_{jf} \end{cases}$$

**Keterangan:**
* Jarak bernilai $0$ jika status binernya sama.
* Jarak bernilai $1$ jika status binernya berbeda.

**Perhitungan:**
* Nilai C1 = No, C2 = Yes
* Karena berbeda, maka $d_2 = 1$

### **C. Jarak Fitur 3: PaymentMethod (Kategori/Nominal)**
**Rumus:**
$$d_{ij}^{(f)} = \begin{cases} 0, & \text{jika } x_{if} = x_{jf} \\ 1, & \text{jika } x_{if} \neq x_{jf} \end{cases}$$

**Keterangan:**
* Jarak bernilai $0$ jika kategorinya persis sama.
* Jarak bernilai $1$ jika kategorinya berbeda (tidak ada tingkatan antar kategori).

**Perhitungan:**
* Nilai C1 = Mailed check, C2 = Electronic check
* Karena berbeda, maka $d_3 = 1$

### **D. Jarak Fitur 4: Contract (Ordinal)**
**Rumus:**
$$d_{ij}^{(f)} = \frac{|z_{if} - z_{jf}|}{R_f}$$

**Keterangan:**
* $z_{if}$: Nilai peringkat (angka) pelanggan pertama.
* $z_{jf}$: Nilai peringkat (angka) pelanggan kedua.
* $R_f$: Rentang peringkat (Peringkat Max - Peringkat Min).

**Perhitungan:**
* Nilai C1 = 1 (Month-to-month), C2 = 3 (Two year), $R_f = 2$
* $d_4 = \frac{|1 - 3|}{2} = \frac{2}{2} = 1$

---

## 3. Total Jarak Campuran C1 vs C2

**Rumus:**
$$Distance(i, j) = \frac{\sum_{f=1}^{n} d_{ij}^{(f)}}{n}$$

**Keterangan:**
* $\sum d_{ij}^{(f)}$: Total jumlah jarak dari seluruh fitur.
* $n$: Jumlah total fitur yang dihitung (dalam kasus ini ada 4).

**Perhitungan:**
* $Total Jarak = \frac{0.5 + 1 + 1 + 1}{4}$
* $Total Jarak = \frac{3.5}{4} = 0.875$

**Kesimpulan:**
Jarak ketidakmiripan antara Pelanggan 1 dan Pelanggan 2 adalah **0.875** (atau 87.5%). Nilai yang mendekati 1 ini menunjukkan bahwa karakteristik kedua pelanggan tersebut sangat berbeda secara keseluruhan.
