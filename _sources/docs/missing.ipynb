# Simulasi k-NN Imputasi pada Dataset Iris

Untuk membuktikan bagaimana k-NN mengisi *missing value*, kita menggunakan dataset **Iris asli**. Kita menyembunyikan nilai `sepal_length` pada **ID 20** (sebagai Target), dan kita akan mencari nilai estimasinya dengan menghitung kedekatannya terhadap seluruh data sebelumnya, yaitu **ID 1 hingga ID 19** (sebagai Referensi).

## 1. Data Target (ID 20)
Data asli `sepal_length` pada ID 20 adalah **5.1**. Untuk simulasi ini, kita anggap nilai tersebut hilang (NaN).

| ID | sepal_length | sepal_width | petal_length | petal_width |
|---|---|---|---|---|
| **20 (Target)** | **`?` (NaN)** | 3.8 | 1.5 | 0.3 |

---

## 2. Perhitungan Jarak ke Populasi Referensi (ID 1 - 19)
Sesuai prinsip k-NN, kita menghitung jarak (Euclidean) dari ID 20 ke **seluruh 19 data referensi**. Perhitungan jarak hanya menggunakan 3 kolom yang tersisa (tanpa `sepal_length`).

**Rumus (Simulasi di Excel):**
`=SQRT(POWER(Width_Target - Width_Ref, 2) + POWER(PetalL_Target - PetalL_Ref, 2) + POWER(PetalW_Target - PetalW_Ref, 2))`

*Catatan: Daripada menuliskan 19 baris perhitungan manual secara panjang lebar di laporan, berikut adalah contoh perhitungan jarak ke beberapa titik:*

* **Jarak ID 20 ke ID 1** $(3.5,\ 1.4,\ 0.2)$:
  $$d_{20\rightarrow1} = \sqrt{(3.8 - 3.5)^2 + (1.5 - 1.4)^2 + (0.3 - 0.2)^2} = \sqrt{0.11} \approx 0.331$$

* **Jarak ID 20 ke ID 11** $(3.7,\ 1.5,\ 0.2)$:
  $$d_{20\rightarrow11} = \sqrt{(3.8 - 3.7)^2 + (1.5 - 1.5)^2 + (0.3 - 0.2)^2} = \sqrt{0.01 + 0 + 0.01} = \sqrt{0.02} \approx 0.141$$

* **Jarak ID 20 ke ID 19** $(3.8,\ 1.7,\ 0.3)$:
  $$d_{20\rightarrow19} = \sqrt{(3.8 - 3.8)^2 + (1.5 - 1.7)^2 + (0.3 - 0.3)^2} = \sqrt{0 + 0.04 + 0} = \sqrt{0.04} = 0.200$$

*(Tahap cross-validation perhitungan dari ID 1 hingga ID 19 dilampirkan menggunakan Microsoft Excel).*

---

## 3. Pemilihan Tetangga Terdekat & Imputasi
Setelah menghitung jarak ke 19 data referensi, algoritma menyortirnya dari jarak terkecil. Jika kita menggunakan **$k=3$**, maka 3 data yang menempati peringkat teratas (paling mirip dengan ID 20) adalah:

1. **Peringkat 1: ID 11** (Jarak = 0.141) $\rightarrow$ `sepal_length` asli = 5.4
2. **Peringkat 2: ID 19** (Jarak = 0.200) $\rightarrow$ `sepal_length` asli = 5.7
3. **Peringkat 3: ID 5** (Jarak = 0.245) $\rightarrow$ `sepal_length` asli = 5.0

**Menghitung Nilai Estimasi (Mean):**
Untuk mengisi *missing value* dari target, kita merata-ratakan nilai `sepal_length` dari ketiga tetangga terdekat tersebut.

$$Nilai\_Imputasi = \frac{5.4 + 5.7 + 5.0}{3}$$
$$Nilai\_Imputasi = \frac{16.1}{3} = \mathbf{5.367}$$

**Kesimpulan Evaluasi:**
Melalui metode k-NN Imputation dengan populasi referensi yang valid (ID 1 - 19) dan nilai tetangga $k=3$, nilai yang hilang pada ID 20 diestimasi menjadi **5.37**. Angka ini sangat realistis dan mendekati nilai aslinya yaitu **5.1**, membuktikan bahwa algoritma ini mampu mengenali pola bunga berdasarkan fitur lebar kelopak dan panjang mahkota secara akurat.
