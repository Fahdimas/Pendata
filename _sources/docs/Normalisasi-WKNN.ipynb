# Preprocessing Data

## Normalisasi Data
Normalisasi data adalah teknik *preprocessing* (pra-pemrosesan) yang digunakan untuk mengubah skala nilai numerik dalam *dataset* agar berada dalam rentang yang seragam dan sebanding, tanpa menghilangkan informasi atau karakteristik asli data tersebut.

Tujuan utamanya adalah untuk mencegah fitur (kolom) yang memiliki rentang nilai sangat besar mendominasi fitur dengan rentang nilai yang kecil saat algoritma *machine learning* melakukan perhitungan jarak.

Sebagai contoh studi kasus, kita akan menggunakan *dataset* berikut untuk mendemonstrasikan perhitungan manual dari setiap metode normalisasi:

| 2 | 3 | 3000000 |
| 3 | 4 | 2000000 |
| 4 | 2 | 2000000 |
| 5 | 3 | 3000000 |
| 6 | 4 | 4000000 |
| 7 | 2 | 3000000 |

### 1. Min-Max Normalization

Min-Max Normalization adalah teknik yang mengubah rentang data agar secara linier masuk ke dalam skala tertentu. Secara *default*, rentang yang digunakan adalah **0 sampai 1**. Angka terkecil di data asli akan menjadi 0, dan angka terbesar akan menjadi 1.

**Rumus:**
$$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

**Keterangan:**
* $X_{norm}$: Nilai hasil normalisasi
* $X$: Nilai asli dari data
* $X_{min}$: Nilai minimum dari atribut tersebut
* $X_{max}$: Nilai maksimum dari atribut tersebut

**Langkah Perhitungan Manual:**
Dari tabel di atas, kita tentukan nilai Min dan Max untuk setiap atribut:
* **IPK:** $Min = 2$, $Max = 4$
* **PO:** $Min = 2000000$, $Max = 4000000$

Mari kita hitung normalisasi untuk **Data ke-1** dan **Data ke-2**:
* **Data 1 (IPK = 2, PO = 2000000):**
  $$IPK_{norm} = \frac{2 - 2}{4 - 2} = \frac{0}{2} = 0$$
  $$PO_{norm} = \frac{2000000 - 2000000}{4000000 - 2000000} = 0$$

* **Data 2 (IPK = 3, PO = 3000000):**
  $$IPK_{norm} = \frac{3 - 2}{4 - 2} = \frac{1}{2} = 0.5$$
  $$PO_{norm} = \frac{3000000 - 2000000}{4000000 - 2000000} = 0.5$$

Jika perhitungan ini diteruskan sampai data ke-7, akan menghasilkan tabel berikut:

| No | IPK_Norm | PO_Norm |
| :---: | :---: | :---: |
| 1 | 0.0 | 0.0 |
| 2 | 0.5 | 0.5 |
| 3 | 1.0 | 0.0 |
| 4 | 0.0 | 0.0 |
| 5 | 0.5 | 0.5 |
| 6 | 1.0 | 1.0 |
| 7 | 0.0 | 0.5 |

**Implementasi Code Python:**

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# Definisi Data
data = np.array([
    [2, 2000000], [3, 3000000], [4, 2000000],
    [2, 2000000], [3, 3000000], [4, 4000000], [2, 3000000]
])

# Proses Min-Max Normalization (Rentang 0-1)
scaler = MinMaxScaler()
hasil_minmax = scaler.fit_transform(data)

print(hasil_minmax)
```

### 2. Z-Score Normalization (Standardization)

Teknik ini mengubah data berdasarkan nilai Rata-rata (*Mean*) dan Standar Deviasinya. Teknik ini sangat cocok untuk data yang memiliki banyak nilai pencilan (*outlier*). Hasil normalisasi ini akan membuat data memiliki rata-rata bernilai 0 dan standar deviasi bernilai 1.

**Rumus Utama Z-Score:**
$$Z = \frac{X - \mu}{\sigma}$$

**Keterangan:**
* $Z$: Nilai hasil normalisasi Z-Score
* $X$: Nilai asli dari data
* $\mu$ (*mu*): Rata-rata dari atribut
* $\sigma$ (*sigma*): Standar Deviasi dari atribut

Untuk menghitung Z-Score secara manual, kita harus mencari nilai $\mu$ dan $\sigma$ terlebih dahulu.

**Langkah 1: Menghitung Rata-rata ($\mu$)**
Rumus rata-rata adalah total semua nilai dibagi jumlah data ($N=7$).
* **Rata-rata IPK:** $$\mu = \frac{2+3+4+2+3+4+2}{7} = \frac{20}{7} \approx 2.857$$
* **Rata-rata PO:** $$\mu = \frac{19000000}{7} \approx 2714285.7$$

**Langkah 2: Menghitung Standar Deviasi ($\sigma$)**
Rumus standar deviasi populasi: $\sigma = \sqrt{\frac{\sum (X - \mu)^2}{N}}$
* **Standar Deviasi IPK:** $$\sigma = \sqrt{\frac{(2-2.857)^2 + (3-2.857)^2 + ...}{7}} \approx 0.833$$
* **Standar Deviasi PO:** $$\sigma \approx 699285.9$$

**Langkah 3: Menghitung Z-Score (Contoh Data ke-1 & ke-2)**
* **Data 1 (IPK = 2, PO = 2000000):**
  $$IPK_{Z} = \frac{2 - 2.857}{0.833} \approx -1.029$$
  $$PO_{Z} = \frac{2000000 - 2714285.7}{699285.9} \approx -1.021$$

* **Data 2 (IPK = 3, PO = 3000000):**
  $$IPK_{Z} = \frac{3 - 2.857}{0.833} \approx 0.171$$
  $$PO_{Z} = \frac{3000000 - 2714285.7}{699285.9} \approx 0.408$$

Hasil keseluruhan Z-Score untuk semua baris:

| No | IPK_ZScore | PO_ZScore |
| :---: | :---: | :---: |
| 1 | -1.029 | -1.021 |
| 2 | 0.171 | 0.408 |
| 3 | 1.372 | -1.021 |
| 4 | -1.029 | -1.021 |
| 5 | 0.171 | 0.408 |
| 6 | 1.372 | 1.838 |
| 7 | -1.029 | 0.408 |

**Implementasi Code Python:**

```python
from sklearn.preprocessing import StandardScaler

# Menggunakan data array yang sama seperti di atas
scaler_z = StandardScaler()
hasil_zscore = scaler_z.fit_transform(data)

print(hasil_zscore)
```

### 3. Decimal Scaling Normalization

Teknik ini memindahkan titik desimal (koma) dari suatu nilai berdasarkan jumlah digit dari angka absolut terbesar di dalam atribut tersebut. Rentang hasil akhirnya selalu berada di antara **-1 sampai 1**.

**Rumus:**
$$X_{norm} = \frac{X}{10^j}$$

**Cara Menentukan Nilai $j$:**
1. Cari nilai absolut terbesar pada atribut.
2. Hitung jumlah digit dari angka tersebut. Jumlah digit itulah yang menjadi nilai $j$.

**Langkah Perhitungan Manual:**
* **Untuk Atribut IPK:** Nilai terbesarnya adalah 4. Angka 4 terdiri dari **1 digit**, sehingga nilai $j = 1$.
* **Untuk Atribut PO:** Nilai terbesarnya adalah 4000000. Angka 4000000 terdiri dari **7 digit**, sehingga nilai $j = 7$. *(Catatan: Nilai pembagi akan menjadi $10^7 = 10000000$)*.

Mari kita hitung untuk **Data ke-1** dan **Data ke-2**:
* **Data 1 (IPK = 2, PO = 2000000):**
  $$IPK_{norm} = \frac{2}{10^1} = 0.2$$
  $$PO_{norm} = \frac{2000000}{10^7} = 0.2$$

* **Data 2 (IPK = 3, PO = 3000000):**
  $$IPK_{norm} = \frac{3}{10^1} = 0.3$$
  $$PO_{norm} = \frac{3000000}{10^7} = 0.3$$

Hasil keseluruhan Decimal Scaling:

| No | IPK_Dec | PO_Dec |
| :---: | :---: | :---: |
| 1 | 0.2 | 0.2 |
| 2 | 0.3 | 0.3 |
| 3 | 0.4 | 0.2 |
| 4 | 0.2 | 0.2 |
| 5 | 0.3 | 0.3 |
| 6 | 0.4 | 0.4 |
| 7 | 0.2 | 0.3 |

**Implementasi Code Python:**
Karena *library* `sklearn` tidak menyediakan fungsi bawaan untuk *Decimal Scaling*, kita dapat membuat fungsinya sendiri dengan Python murni.

```python
import numpy as np

# Membuat fungsi Decimal Scaling
def decimal_scaling(data_kolom):
    max_val = np.max(np.abs(data_kolom))
    j = len(str(int(max_val)))
    return data_kolom / (10**j)

# Mengambil kolom secara terpisah
ipk_col = data[:, 0]
po_col = data[:, 1]

# Menerapkan fungsi
hasil_ipk_dec = decimal_scaling(ipk_col)
hasil_po_dec = decimal_scaling(po_col)

# Menggabungkan hasil
hasil_decimal = np.column_stack((hasil_ipk_dec, hasil_po_dec))
print(hasil_decimal)
```

## Penyelesaian Missing Values dengan WKNN

Setelah memahami konsep dari ketiga normalisasi data di atas, kita akan mengimplementasikannya secara langsung untuk menyelesaikan kasus *missing value* (data yang hilang) menggunakan algoritma **Weighted K-Nearest Neighbors (WKNN)**.

WKNN dipilih karena algoritma ini memberikan **bobot** (kemiripan) yang lebih besar pada tetangga yang jaraknya lebih dekat dengan data target. Semakin dekat jaraknya, semakin besar pengaruh tetangga tersebut terhadap hasil prediksi.

Sebagai studi kasus, kita akan mencari nilai yang hilang ("?") pada kolom **JML** di baris ke-7 dari tabel berikut:

| No | IPK | PO | JML |
| :---: | :---: | :---: | :---: |
| 1 | 2 | 2000000 | 2 |
| 2 | 3 | 3000000 | 3 |
| 3 | 4 | 2000000 | 2 |
| 4 | 2 | 2000000 | 3 |
| 5 | 3 | 3000000 | 2 |
| 6 | 4 | 4000000 | 3 |
| **7 (Target)** | **2** | **3000000** | **?** |

---

### Tahap 1: Normalisasi Data (Min-Max)
Sebelum menghitung kemiripan antardata, kita wajib menyamakan skala datanya terlebih dahulu menggunakan **Normalisasi Min-Max** (rentang 0 sampai 1).

Berdasarkan rumus Min-Max: $$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

Kita tentukan nilai Min dan Max dari baris 1 sampai 6:
* **IPK:** $Min = 2$, $Max = 4$
* **PO:** $Min = 2000000$, $Max = 4000000$
* **JML:** $Min = 2$, $Max = 3$

*(Perhitungan dilakukan ke seluruh sel data, sehingga menghasilkan tabel data yang sudah dinormalisasi berikut ini):*

| No | IPK_Norm | PO_Norm | JML_Norm |
| :---: | :---: | :---: | :---: |
| 1 | 0.0 | 0.0 | 0.0 |
| 2 | 0.5 | 0.5 | 1.0 |
| 3 | 1.0 | 0.0 | 0.0 |
| 4 | 0.0 | 0.0 | 1.0 |
| 5 | 0.5 | 0.5 | 0.0 |
| 6 | 1.0 | 1.0 | 1.0 |
| **7 (Target)**| **0.0** | **0.5** | **?** |

---

### Tahap 2: Hitung Jarak dan Kemiripan ($s_i$)
Tahap selanjutnya adalah mencari seberapa mirip Data Target (baris 7) dengan data-data lainnya (baris 1 sampai 6).

* **Data Target ($X_{target}$):** IPK = 0.0, PO = 0.5.
* **Rumus Kemiripan ($s_i$):** Kita menggunakan kebalikan dari jarak Euclidean yang dikuadratkan:
  $$s_i = \frac{1}{(x_{target} - x_i)^2 + (y_{target} - y_i)^2}$$

Mari kita hitung nilai kemiripan untuk semua data latih ($K=6$):
* **Data 1:** Jarak Kuadrat = $(0.0 - 0.0)^2 + (0.5 - 0.0)^2 = 0.25$
  Kemiripan ($s_1$) = $\frac{1}{0.25} = \mathbf{4.0}$
* **Data 2:** Jarak Kuadrat = $(0.0 - 0.5)^2 + (0.5 - 0.5)^2 = 0.25$
  Kemiripan ($s_2$) = $\frac{1}{0.25} = \mathbf{4.0}$
* **Data 3:** Jarak Kuadrat = $(0.0 - 1.0)^2 + (0.5 - 0.0)^2 = 1.25$
  Kemiripan ($s_3$) = $\frac{1}{1.25} = \mathbf{0.8}$
* **Data 4:** Jarak Kuadrat = $(0.0 - 0.0)^2 + (0.5 - 0.0)^2 = 0.25$
  Kemiripan ($s_4$) = $\frac{1}{0.25} = \mathbf{4.0}$
* **Data 5:** Jarak Kuadrat = $(0.0 - 0.5)^2 + (0.5 - 0.5)^2 = 0.25$
  Kemiripan ($s_5$) = $\frac{1}{0.25} = \mathbf{4.0}$
* **Data 6:** Jarak Kuadrat = $(0.0 - 1.0)^2 + (0.5 - 1.0)^2 = 1.25$
  Kemiripan ($s_6$) = $\frac{1}{1.25} = \mathbf{0.8}$

---

### Tahap 3: Hitung Estimasi (Weighted Average)
Setelah mendapatkan bobot kemiripan dari setiap tetangga, kita akan memprediksi nilai JML Normalisasi menggunakan metode *Weighted Average* (Rata-rata Tertimbang).

**Rumus:** $$Prediksi = \frac{\text{Total Perkalian (Kemiripan} \times \text{JML\_Norm)}}{\text{Total Kemiripan}}$$

**A. Menghitung Pembilang:**
Kita kalikan kemiripan ($s_i$) dengan target JML_Norm masing-masing baris:
* Data 1: $4.0 \times 0.0 = 0.0$
* Data 2: $4.0 \times 1.0 = 4.0$
* Data 3: $0.8 \times 0.0 = 0.0$
* Data 4: $4.0 \times 1.0 = 4.0$
* Data 5: $4.0 \times 0.0 = 0.0$
* Data 6: $0.8 \times 1.0 = 0.8$
**Total Pembilang:** $0.0 + 4.0 + 0.0 + 4.0 + 0.0 + 0.8 = \mathbf{8.8}$

**B. Menghitung Penyebut (Total Kemiripan):**
* **Total Penyebut:** $4.0 + 4.0 + 0.8 + 4.0 + 4.0 + 0.8 = \mathbf{17.6}$

**C. Hasil Estimasi Normalisasi:**
$$Prediksi_{norm} = \frac{8.8}{17.6} = \mathbf{0.5}$$

---

### Tahap 4: Denormalisasi (Nilai Asli)
Karena hasil prediksi (0.5) masih dalam bentuk normalisasi, kita harus mengembalikannya ke bentuk aslinya untuk mengisi tabel.

**Rumus Denormalisasi:**
$$X_{asli} = X_{norm} \times (X_{max} - X_{min}) + X_{min}$$

Kita masukkan nilai prediksi dan batas Min/Max dari kolom JML asli:
* $X_{norm} = 0.5$
* $X_{min} = 2$
* $X_{max} = 3$

$$JML_{asli} = 0.5 \times (3 - 2) + 2$$
$$JML_{asli} = 0.5 \times 1 + 2 = \mathbf{2.5}$$

> **Kesimpulan:** Nilai *missing value* pada kolom JML untuk baris ke-7 telah berhasil diselesaikan. Estimasi nilai yang paling tepat menggunakan metode WKNN adalah **2.5**.

### Implementasi Code WKNN Python
Berikut adalah *full code* menggunakan *library* `scikit-learn` untuk menyelesaikan *missing value* dengan algoritma WKNN, persis seperti hitungan manual di atas.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor

# 1. Definisi Data
data = pd.DataFrame({
    'IPK': [2, 3, 4, 2, 3, 4, 2],
    'PO': [2000000, 3000000, 2000000, 2000000, 3000000, 4000000, 3000000],
    'JML': [2.0, 3.0, 2.0, 3.0, 2.0, 3.0, np.nan]
})

# 2. Proses Normalisasi Min-Max
scaler_features = MinMaxScaler()
scaler_target = MinMaxScaler()

# Pisahkan Data Latih (baris 1-6) dan Target yang hilang (baris 7)
df_train = data.iloc[:6]
df_test = data.iloc[6:]

# Terapkan Normalisasi
X_train = scaler_features.fit_transform(df_train[['IPK', 'PO']])
X_test = scaler_features.transform(df_test[['IPK', 'PO']])
y_train = scaler_target.fit_transform(df_train[['JML']])

# 3. Prediksi dengan WKNN (Rumus Jarak Kuadrat)
# Mendefinisikan fungsi bobot khusus: 1 / (jarak^2)
def custom_weights(distances):
    # Menggunakan epsilon kecil untuk mencegah error pembagian dengan 0
    return 1 / (distances**2 + 1e-10)

# Inisialisasi model WKNN (Menggunakan K=6 agar semua data latih dihitung)
wknn = KNeighborsRegressor(n_neighbors=6, weights=custom_weights, metric='euclidean')
wknn.fit(X_train, y_train)

# 4. Lakukan prediksi dan denormalisasi
prediksi_norm = wknn.predict(X_test)
prediksi_asli = scaler_target.inverse_transform(prediksi_norm)

print("=== HASIL PREDIKSI WKNN ===")
print(f"Prediksi JML (Normalisasi): {prediksi_norm[0][0]:.1f}")
print(f"Hasil Akhir Prediksi JML (Denormalisasi): {prediksi_asli[0][0]:.1f}")
```
