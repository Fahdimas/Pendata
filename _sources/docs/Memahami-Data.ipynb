# Memahami Karakteristik Data

Halaman ini membahas mengenai teknik statistik yang digunakan untuk membedah karakteristik serta melihat kualitas dari data yang akan diolah dalam proses *data mining*.

## Pengertian Metrik Statistik Data

### 1. Mean (Rata-Rata)
Mean adalah angka yang mewakili nilai rata-rata dari seluruh kumpulan data yang kita miliki. Cara mendapatkannya adalah dengan menjumlahkan semua angka dalam dataset lalu membaginya dengan jumlah total datanya.

**Rumus Mean:**
$$\bar{x} = \frac{\sum_{i=1}^{n} x_i}{n}$$

**Keterangan:**
* $\bar{x}$ : Titik rata-rata (Mean).
* $x_i$ : Nilai data pada urutan tertentu.
* $n$ : Total jumlah sampel data.
* $\sum$ : Simbol untuk menjumlahkan keseluruhan nilai.

### 2. Median (Nilai Tengah)
Median adalah titik pusat dataset setelah semua angkanya disusun secara berurutan dari yang terkecil hingga terbesar. Nilai ini dianggap lebih stabil karena tidak mudah dipengaruhi oleh angka ekstrem atau pencilan (*outlier*).

**Rumus untuk Data Tunggal:**
* Jika jumlah data ($N$) Ganjil: $Me = x_{\frac{N+1}{2}}$
* Jika jumlah data ($N$) Genap: $Me = \frac{x_{\frac{N}{2}} + x_{\frac{N}{2}+1}}{2}$

**Keterangan:**
* $Me$ : Nilai tengah (Median).
* $x$ : Posisi data tertentu.
* $N$ : Banyaknya jumlah data.

### 3. Modus/Mode (Nilai Terpopuler)
Modus adalah angka atau kategori yang frekuensi kemunculannya paling tinggi di dalam dataset.

**Hubungan Empiris (untuk distribusi yang tidak seimbang):**
$$\text{Mean} - \text{Modus} \approx 3(\text{Mean} - \text{Median})$$

### 4. Rentang (Range) dan Jarak Interkuartil (IQR)
Dua metrik ini sangat berguna untuk melihat sebaran data secara luas serta mendeteksi apakah ada data yang tidak wajar.

**Rumus Range:**
$$R = X_{max} - X_{min}$$

**Rumus Interkuartil:**
$$IQR = Q_3 - Q_1$$

**Keterangan:**
* $R$ : Selisih atau rentang data.
* $X_{max}$ : Angka paling tinggi di dataset.
* $X_{min}$ : Angka paling rendah di dataset.
* $IQR$ : Jarak antara kuartil atas dan bawah.
* $Q_1$ & $Q_3$ : Batas data 25% (bawah) dan 75% (atas).

### 5. Variansi dan Standar Deviasi
Dua nilai ini menunjukkan seberapa jauh titik-titik data menyimpang dari nilai rata-ratanya.

**Rumus Standar Deviasi ($\sigma$):**
$$\sigma = \sqrt{\frac{\sum_{i=1}^{N} (x_i - \bar{x})^2}{N}}$$

**Keterangan:**
* $\sigma$ : Simpangan baku atau standar deviasi.
* $x_i$ : Nilai tiap individu data.
* $\bar{x}$ : Rata-rata dari data tersebut.
* $N$ : Jumlah keseluruhan data.

### 6. Skewness (Kemencengan)
Skewness digunakan untuk mengukur apakah bentuk sebaran data kita miring ke satu sisi atau seimbang di tengah seperti kurva normal.

**Rumus Koefisien Pearson:**
$$sk = \frac{3(\bar{x} - Me)}{s}$$

**Keterangan:**
* $sk$ : Angka kemencengan (*Skewness*).
* $\bar{x}$ : Nilai rata-rata.
* $Me$ : Nilai median.
* $s$ : Standar deviasi dari sampel tersebut.
