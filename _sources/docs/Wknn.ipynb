# Simulasi W-KNN Imputasi pada Data Numerik (JMK)

Untuk membuktikan bagaimana algoritma *Weighted k-Nearest Neighbors* (W-KNN) mengisi kekosongan data (*missing value*), kita mengambil **6 baris data** dari dataset preprocessing. 

Kita menyimulasikan skenario di mana nilai `JMK` pada baris ke-7 hilang (sebagai Target). Kita akan mencari estimasi nilainya dengan menghitung kedekatannya terhadap 6 data di atasnya (sebagai Referensi) menggunakan tabel yang telah **dinormalisasi**.

## 1. Data Normalisasi & Skenario Missing Value
Berdasarkan tabel normalisasi, nilai `JMK` pada baris ke-7 sengaja kita hilangkan (NaN) untuk diuji. Atribut prediktor yang digunakan untuk menghitung kedekatan adalah `IPK` dan `PO`.

| No | IPK | PO | JMK |
|---|---|---|---|
| **1 (Ref)** | 0 | 0 | 0 |
| **2 (Ref)** | 0.5 | 0.5 | 1 |
| **3 (Ref)** | 1 | 0 | 0 |
| **4 (Ref)** | 0 | 0 | 1 |
| **5 (Ref)** | 0.5 | 0.5 | 0 |
| **6 (Ref)** | 1 | 1 | 1 |
| **7 (Target)** | 0 | 0.5 | **`?` (NaN)** |

---

## 2. Perhitungan Jarak Euclidean ke Populasi Referensi
Kita menghitung metrik jarak (Euclidean Distance) dari target ke 6 referensi menggunakan atribut `IPK` dan `PO`.

**Rumus:**
$$d = \sqrt{(IPK_{target} - IPK_{ref})^2 + (PO_{target} - PO_{ref})^2}$$

* **Jarak ke No. 1** $(0,\ 0)$:
  $$d_1 = \sqrt{(0 - 0)^2 + (0.5 - 0)^2} = \sqrt{0 + 0.25} = \mathbf{0.5}$$

* **Jarak ke No. 2** $(0.5,\ 0.5)$:
  $$d_2 = \sqrt{(0 - 0.5)^2 + (0.5 - 0.5)^2} = \sqrt{0.25 + 0} = \mathbf{0.5}$$

* **Jarak ke No. 3** $(1,\ 0)$:
  $$d_3 = \sqrt{(0 - 1)^2 + (0.5 - 0)^2} = \sqrt{1 + 0.25} \approx \mathbf{1.118}$$

* **Jarak ke No. 4** $(0,\ 0)$:
  $$d_4 = \sqrt{(0 - 0)^2 + (0.5 - 0)^2} = \sqrt{0 + 0.25} = \mathbf{0.5}$$

* **Jarak ke No. 5** $(0.5,\ 0.5)$:
  $$d_5 = \sqrt{(0 - 0.5)^2 + (0.5 - 0.5)^2} = \sqrt{0.25 + 0} = \mathbf{0.5}$$

* **Jarak ke No. 6** $(1,\ 1)$:
  $$d_6 = \sqrt{(0 - 1)^2 + (0.5 - 1)^2} = \sqrt{1 + 0.25} \approx \mathbf{1.118}$$

---

## 3. Pemilihan Tetangga Terdekat & Perhitungan Bobot
Setelah menghitung jarak, algoritma mendapati **4 data** dengan jarak terdekat yang sama kuat ($d = 0.5$), yaitu data No. 1, 2, 4, dan 5. 

Dalam metode W-KNN, kita perlu memberikan bobot ($w$) pada masing-masing tetangga terdekat tersebut berdasarkan jaraknya.

**Rumus Bobot:**
$$w = \frac{1}{d^2}$$

Karena jarak keempat tetangga sama (0.5), maka bobotnya juga sama:
* $w_1, w_2, w_4, w_5 = \frac{1}{0.5^2} = \frac{1}{0.25} = \mathbf{4}$

**Total Bobot** $(\sum w) = 4 + 4 + 4 + 4 = \mathbf{16}$

---

## 4. Imputasi Nilai JMK
Langkah terakhir adalah menghitung nilai estimasi W-KNN dengan mengalikan bobot dan nilai JMK normalisasi dari masing-masing tetangga, lalu membaginya dengan total bobot.
* JMK No. 1 = 0
* JMK No. 2 = 1
* JMK No. 4 = 1
* JMK No. 5 = 0

**Menghitung Nilai Estimasi W-KNN:**
$$Nilai\_Imputasi = \frac{(w_1 \times JMK_1) + (w_2 \times JMK_2) + (w_4 \times JMK_4) + (w_5 \times JMK_5)}{\sum w}$$

$$Nilai\_Imputasi = \frac{(4 \times 0) + (4 \times 1) + (4 \times 1) + (4 \times 0)}{16}$$

$$Nilai\_Imputasi = \frac{0 + 4 + 4 + 0}{16} = \frac{8}{16} = \mathbf{0.5}$$

### **Kesimpulan Evaluasi:**
Melalui metode W-KNN Imputation, nilai `JMK` normalisasi yang hilang berhasil diestimasi menjadi **0.5**. Jika didenormalisasi kembali ke rentang nilai aslinya (di mana 0 = 2 dan 1 = 3), posisinya berada tepat di angka **2.5**. Mengingat atribut JMK merupakan data diskrit, nilai ini sangat rasional dan siap dibulatkan menjadi **3** atau **2** sesuai dengan kebijakan pembulatan yang diterapkan pada studi kasus.
