# Pengantar Data Mining

Halaman ini menjelaskan konsep dasar Data Mining, mengapa ia penting, dan bagaimana prosesnya dilakukan mengikut piawaian industri.

## 1. Definisi Data Mining
Data Mining adalah proses mengekstraksi informasi atau pola yang menarik dan sebelumnya tidak diketahui dari sekumpulan data yang besar. Ia sering juga disebut sebagai **Knowledge Discovery in Databases (KDD)**.

## 2. Mengapa Data Mining Penting?
* **Ledakan Data**: Kita mempunyai terlalu banyak data tetapi kekurangan maklumat berharga.
* **Pengambilan Keputusan**: Membantu organisasi membuat keputusan berdasarkan bukti data, bukan sekadar firasat.
* **Analisis Trend**: Mengenal pasti corak tingkah laku pelanggan atau trend pasaran di masa depan.

## 3. Proses KDD (Knowledge Discovery in Databases)
Proses mendapatkan pengetahuan daripada data melibatkan beberapa langkah:
1. **Data Cleaning**: Membuang data yang tidak konsisten atau "noise".
2. **Data Integration**: Menggabungkan data dari pelbagai sumber.
3. **Data Selection**: Memilih data yang relevan dengan analisis.
4. **Data Transformation**: Mengubah data ke dalam bentuk yang sesuai untuk diproses.
5. **Data Mining**: Menggunakan algoritma untuk mencari pola.
6. **Pattern Evaluation**: Menilai pola yang ditemui.
7. **Knowledge Presentation**: Mempersembahkan hasil dalam bentuk visual yang mudah difahami.

## 4. Metodologi CRISP-DM
Siklus kerja utama yang digunakan dalam projek ini adalah **CRISP-DM** (*Cross-Industry Standard Process for Data Mining*).

![Siklus CRISP-DM](../docs/crisp.png)

Siklus ini terdiri daripada 6 tahap utama:
* **Business Understanding**: Menetapkan tujuan projek.
* **Data Understanding**: Memeriksa kualiti dan ciri-ciri data.
* **Data Preparation**: Membersihkan dan menukar format data.
* **Modeling**: Memilih teknik seperti klasifikasi atau clustering.
* **Evaluation**: Memastikan model menepati objektif awal.
* **Deployment**: Menggunakan hasil analisis untuk kegunaan nyata.

## 5. Fungsi Utama Data Mining
* **Klasifikasi**: Menentukan kategori bagi data baru (Contoh: Menentukan spesies bunga Iris).
* **Clustering**: Mengelompokkan data berdasarkan kemiripan tanpa kategori awal.
* **Asosiasi**: Menemukan hubungan antara item (Contoh: Produk yang sering dibeli bersama).
* **Regresi**: Memprediksi nilai numerik (Contoh: Ramalan harga rumah).
