### Tahap 2: Menghitung WKNN (Mencari "?" pada Tabel Normalisasi)

Kita gunakan tabel yang **sudah dinormalisasi** untuk memprediksi nilai JML baris ke-7. Target JML yang digunakan juga yang sudah dinormalisasi.

* **Data Target (Baris 7):** $X_{target} = (0, 0.5)$
* **Data Latih (Baris 1-6):**
    * $X_1 = (0, 0)$, $Y_1 = 0$
    * $X_2 = (0.5, 0.5)$, $Y_2 = 1$
    * $X_3 = (1, 0)$, $Y_3 = 0$
    * $X_4 = (0, 0)$, $Y_4 = 1$
    * $X_5 = (0.5, 0.5)$, $Y_5 = 0$
    * $X_6 = (1, 1)$, $Y_6 = 1$

**Langkah 1: Hitung Jarak Euclidean ($d$)**
Hitung jarak dari $X_{target}$ ke semua data latih: $d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$
* $d_1 = \sqrt{(0 - 0)^2 + (0 - 0.5)^2} = \sqrt{0 + 0.25} = 0.5$
* $d_2 = \sqrt{(0.5 - 0)^2 + (0.5 - 0.5)^2} = \sqrt{0.25 + 0} = 0.5$
* $d_3 = \sqrt{(1 - 0)^2 + (0 - 0.5)^2} = \sqrt{1 + 0.25} \approx 1.118$
* $d_4 = \sqrt{(0 - 0)^2 + (0 - 0.5)^2} = \sqrt{0 + 0.25} = 0.5$
* $d_5 = \sqrt{(0.5 - 0)^2 + (0.5 - 0.5)^2} = \sqrt{0.25 + 0} = 0.5$
* $d_6 = \sqrt{(1 - 0)^2 + (1 - 0.5)^2} = \sqrt{1 + 0.25} \approx 1.118$

**Langkah 2: Tentukan Nilai K dan Hitung Bobot ($w$)**
Terdapat **4 data** dengan jarak terdekat yang sama persis (Baris 1, 2, 4, 5 dengan $d = 0.5$). Maka kita gunakan $K=4$.
Hitung bobot dengan rumus $w = \frac{1}{d}$:
* $w_1 = \frac{1}{0.5} = 2$
* $w_2 = \frac{1}{0.5} = 2$
* $w_4 = \frac{1}{0.5} = 2$
* $w_5 = \frac{1}{0.5} = 2$

**Langkah 3: Hitung Prediksi (Weighted Average)**
Kalikan bobot dengan target JML Normalisasi ($Y$), lalu bagi dengan total bobot.
$$Prediksi = \frac{\sum (w_i \times Y_i)}{\sum w_i}$$
$$Prediksi = \frac{(w_1 \times Y_1) + (w_2 \times Y_2) + (w_4 \times Y_4) + (w_5 \times Y_5)}{w_1 + w_2 + w_4 + w_5}$$
$$Prediksi = \frac{(2 \times 0) + (2 \times 1) + (2 \times 1) + (2 \times 0)}{2 + 2 + 2 + 2}$$
$$Prediksi = \frac{0 + 2 + 2 + 0}{8} = \frac{4}{8} = 0.5$$

> **Hasil:** Tanda "?" pada kolom **Normalisasi JML (kanan)** adalah **0.5**.

### Tahap 3: Denormalisasi (Mencari "?" pada Tabel JML Asli)

Karena hasil prediksi dari WKNN tadi bentuknya masih normalisasi ($0.5$), kita kembalikan ke bentuk angka numerik asli untuk mengisi tabel sebelah kiri.

**Rumus Denormalisasi:**
$$X_{asli} = X_{norm} \times (X_{max} - X_{min}) + X_{min}$$

Kita masukkan nilai prediksi dan nilai Min/Max dari data JML asli:
* $X_{norm} = 0.5$
* $X_{min} = 2$
* $X_{max} = 3$

$$JML_{asli} = 0.5 \times (3 - 2) + 2$$
$$JML_{asli} = 0.5 \times (1) + 2$$
$$JML_{asli} = 0.5 + 2 = 2.5$$

> **Hasil:** Tanda "?" pada kolom **JML asli (kiri)** adalah **2.5**.

Code Menggunakan Normalisasi Min-Max

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics.pairwise import euclidean_distances

# ==========================================
# 1. DEFINISI DATA
# ==========================================
# Baris ke-7 pada kolom JML kita isi dengan np.nan (kosong/missing value)
data = pd.DataFrame({
    'IPK': [2, 3, 4, 2, 3, 4, 2],
    'PO': [2000000, 3000000, 2000000, 2000000, 3000000, 4000000, 3000000],
    'JML': [2.0, 3.0, 2.0, 3.0, 2.0, 3.0, np.nan] 
})

print("--- 1. DATA ASLI ---")
print(data)
print("-" * 30, "\n")

# Pisahkan Fitur (X) dan Target (Y)
X = data[['IPK', 'PO']]
y = data['JML']

# ==========================================
# 2. PROSES NORMALISASI MIN-MAX
# ==========================================
scaler = MinMaxScaler()
# Fit dan transform seluruh data X (baris 1-7)
X_norm_array = scaler.fit_transform(X) 

# Masukkan ke DataFrame baru khusus agar rapi saat di-print
df_norm = pd.DataFrame(X_norm_array, columns=['IPK_Norm', 'PO_Norm'])

print("--- 2. HASIL NORMALISASI MIN-MAX (FITUR X) ---")
print(df_norm)
print("-" * 30, "\n")

# Pisahkan Data Latih (baris 1-6) dan Target yang mau ditebak (baris 7)
X_train_norm = X_norm_array[:6]    # Ambil baris index 0 sampai 5
y_train = y[:6]                    # Target JML asli baris 1-6
X_target_norm = X_norm_array[6:]   # Ambil baris index 6 (baris ke-7)

# ==========================================
# 3. MENGHITUNG JARAK EUCLIDEAN
# ==========================================
# Hitung jarak dari baris ke-7 ke semua data latih (baris 1-6)
jarak = euclidean_distances(X_target_norm, X_train_norm)[0]

print("--- 3. JARAK EUCLIDEAN (TARGET KE DATA LATIH) ---")
for i, d in enumerate(jarak):
    print(f"Jarak ke Baris {i+1} (d{i+1}): {d:.4f}")
print("-" * 30, "\n")

# ==========================================
# 4. PREDIKSI DENGAN WKNN
# ==========================================
# Inisialisasi model WKNN (k=4, weights='distance' agar jadi WKNN)
wknn = KNeighborsRegressor(n_neighbors=4, weights='distance')
wknn.fit(X_train_norm, y_train)

# Lakukan prediksi
prediksi_jml = wknn.predict(X_target_norm)

print("--- 4. HASIL AKHIR PREDIKSI WKNN ---")
print(f"Prediksi nilai JML (Asli) untuk baris ke-7 adalah : {prediksi_jml[0]}")

# Opsional: Jika ingin menampilkan nilai Normalisasi JML-nya juga
jml_min = y_train.min()
jml_max = y_train.max()
jml_norm_pred = (prediksi_jml[0] - jml_min) / (jml_max - jml_min)

print(f"Nilai Normalisasi JML untuk baris ke-7 adalah   : {jml_norm_pred}")
print("-" * 30)
```

### 1. Min-Max Normalization (Scaling)

Metode ini paling sering dipakai untuk mengubah rentang data menjadi skala **0 sampai 1**. Angka terkecil akan menjadi 0, dan angka terbesar akan menjadi 1.

**Rumus:**
$$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

**Keterangan:**
* $X_{norm}$: Nilai data baru setelah dinormalisasi (hasil).
* $X$: Nilai data asli yang sedang dihitung.
* $X_{min}$: Nilai paling kecil dari seluruh data.
* $X_{max}$: Nilai paling besar dari seluruh data.

**Contoh Perhitungan Manual:**
Misal kita punya data Umur: **20, 30, 50**.
* Cari $X_{min} = 20$
* Cari $X_{max} = 50$

Sekarang kita masukkan ke rumus satu per satu:
* Untuk umur **20**: $$X_{norm} = \frac{20 - 20}{50 - 20} = \frac{0}{30} = 0$$
* Untuk umur **30**: $$X_{norm} = \frac{30 - 20}{50 - 20} = \frac{10}{30} \approx 0.333$$
* Untuk umur **50**: $$X_{norm} = \frac{50 - 20}{50 - 20} = \frac{30}{30} = 1$$

Hasil akhir datanya menjadi: **0, 0.333, 1**.

**Pembuktian dengan Code (Sklearn & Manual):**

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({'Umur': [20, 30, 50]})

# 1. Bikin Fungsi Manual
def minmax_manual(x):
    return (x - x.min()) / (x.max() - x.min())

df['Manual'] = minmax_manual(df['Umur'])

# 2. Pakai Library Sklearn
scaler = MinMaxScaler()
df['Sklearn'] = scaler.fit_transform(df[['Umur']])

print(df)
```

### 2. Z-Score Normalization (Standardization)

Metode ini mengubah data berdasarkan rata-rata dan sebaran datanya. Hasil akhirnya akan membuat data memiliki **rata-rata = 0** dan **standar deviasi = 1**. Data yang nilainya di bawah rata-rata akan bernilai minus, dan yang di atas rata-rata bernilai plus.

**Rumus:**
$$Z = \frac{X - \mu}{\sigma}$$

**Keterangan:**
* $Z$: Nilai data baru setelah dinormalisasi (Z-score).
* $X$: Nilai data asli yang sedang dihitung.
* $\mu$ (*mu*): Rata-rata (*Mean*) dari seluruh data.
* $\sigma$ (*sigma*): Standar Deviasi dari seluruh data.

**Contoh Perhitungan Manual:**
Misal kita punya data Nilai: **2, 4, 6**.
* Cari Rata-rata ($\mu$): $$(2 + 4 + 6) / 3 = 4$$
* Cari Varian: $$((2-4)^2 + (4-4)^2 + (6-4)^2) / 3 = (4 + 0 + 4) / 3 = 2.667$$
* Cari Standar Deviasi ($\sigma$): $$\sqrt{2.667} \approx 1.633$$

Sekarang kita hitung Z-score-nya:
* Untuk nilai **2**: $$Z = \frac{2 - 4}{1.633} = \frac{-2}{1.633} \approx -1.22$$
* Untuk nilai **4**: $$Z = \frac{4 - 4}{1.633} = \frac{0}{1.633} = 0$$
* Untuk nilai **6**: $$Z = \frac{6 - 4}{1.633} = \frac{2}{1.633} \approx 1.22$$

Hasil akhir datanya menjadi: **-1.22, 0, 1.22**.

**Pembuktian dengan Code (Sklearn & Manual):**

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({'Nilai': [2, 4, 6]})

# 1. Bikin Fungsi Manual
def zscore_manual(x):
    # ddof=0 agar sama persis dengan perhitungan populasi Sklearn
    return (x - x.mean()) / x.std(ddof=0) 

df['Manual'] = zscore_manual(df['Nilai'])

# 2. Pakai Library Sklearn
scaler = StandardScaler()
df['Sklearn'] = scaler.fit_transform(df[['Nilai']])

print(df)
```

### 3. Decimal Scaling Normalization

Metode ini sangat praktis jika datanya berupa angka ribuan/jutaan. Kita hanya perlu menggeser koma desimal ke kiri berdasarkan jumlah digit dari angka terbesarnya.

**Rumus:**
$$X_{norm} = \frac{X}{10^j}$$

**Keterangan:**
* $X_{norm}$: Nilai data baru setelah dinormalisasi.
* $X$: Nilai data asli yang sedang dihitung.
* $j$: Jumlah digit dari angka *absolut* (positif) yang paling besar di dalam data.
* $10^j$: Angka pembaginya.

**Contoh Perhitungan Manual:**
Misal kita punya data Harga: **150, 2500, -8500**.
* Cari angka absolut terbesar: Antara 150, 2500, dan 8500 (minus diabaikan). Yang terbesar adalah **8500**.
* Cari jumlah digit ($j$): Angka 8500 memiliki **4 digit**. Maka $j = 4$.
* Angka pembagi ($10^j$): $10^4 = 10000$.

Sekarang kita bagi setiap data dengan 10.000:
* Untuk harga **150**: $$X_{norm} = \frac{150}{10000} = 0.015$$
* Untuk harga **2500**: $$X_{norm} = \frac{2500}{10000} = 0.25$$
* Untuk harga **-8500**: $$X_{norm} = \frac{-8500}{10000} = -0.85$$

Hasil akhir datanya menjadi: **0.015, 0.25, -0.85**.

**Pembuktian dengan Code (Fungsi Manual):**
*(Catatan: Sklearn tidak menyediakan fungsi bawaan untuk Decimal Scaling karena sangat sederhana, jadi kita murni buat fungsinya sendiri).*

```python
import pandas as pd

df = pd.DataFrame({'Harga': [150, 2500, -8500]})

# Bikin Fungsi Manual
def decimal_scaling(x):
    max_abs = x.abs().max()       # Cari angka absolut terbesar (8500)
    j = len(str(int(max_abs)))    # Hitung digitnya (4)
    return x / (10**j)            # Bagi data dengan 10^j

df['Manual'] = decimal_scaling(df['Harga'])

print(df)
```
