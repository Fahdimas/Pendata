# Evaluation

Tahap **Evaluation** bertujuan untuk mengukur sejauh mana model yang telah dibangun memenuhi tujuan bisnis atau penelitian yang telah ditetapkan di awal. Kita akan melihat performa model melalui berbagai metrik evaluasi.

## 1. Metrik Evaluasi
Beberapa metrik yang digunakan untuk mengukur kinerja model klasifikasi antara lain:
* **Accuracy**: Persentase total prediksi benar dibandingkan dengan keseluruhan data.
* **Precision**: Rasio prediksi benar positif dibandingkan dengan keseluruhan hasil yang diprediksi positif.
* **Recall**: Rasio prediksi benar positif dibandingkan dengan keseluruhan data yang benar-benar positif.
* **F1-Score**: Rata-rata harmonik antara Precision dan Recall, memberikan gambaran keseimbangan performa model.

## 2. Analisis Confusion Matrix
*Confusion Matrix* digunakan untuk melihat detail prediksi mana yang benar dan mana yang salah (salah klasifikasi) untuk setiap spesies bunga Iris. Melalui tabel ini, kita bisa mengetahui apakah model sering tertukar dalam membedakan spesies *Versicolor* dan *Virginica*.

## 3. Hasil Pengujian (Orange)
Pada aplikasi Orange, hasil evaluasi dapat dilihat pada widget **Test and Score**. Nilai akurasi yang mendekati 1.0 (atau 100%) menunjukkan bahwa model memiliki kemampuan prediksi yang sangat baik pada dataset Iris ini.
