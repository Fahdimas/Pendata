## 1. Landasan Teori: Random Forest
Dalam tugas ini, saya menerapkan algoritma **Random Forest**, yaitu metode *ensemble learning* yang membangun banyak pohon keputusan (Decision Trees) secara acak untuk mendapatkan hasil prediksi yang lebih stabil dan akurat.

Beberapa konsep kunci yang saya terapkan adalah:
* **Bootstrapping:** Proses pengambilan sampel data secara acak dengan pengembalian untuk melatih setiap pohon individu.
* **Feature Randomness:** Setiap pohon hanya melihat subset acak dari fitur yang ada, sehingga pohon-pohon tersebut tidak saling berkorelasi.
* **Majority Voting:** Untuk klasifikasi, hasil akhir diambil dari suara terbanyak dari seluruh pohon.

## 3. Simulasi Perhitungan Manual (Gini Impurity)
Untuk memahami bagaimana pohon dalam Random Forest membelah data, saya melakukan perhitungan manual **Gini Impurity** pada akar pohon (*root node*) menggunakan dataset "Play Tennis" (14 baris).

### Rumus Gini Impurity:
$$Gini(S) = 1 - \sum_{i=1}^{n} (p_i)^2$$

**Keterangan Simbol:**
* **$Gini(S)$**: Nilai ketidakmurnian data. Semakin mendekati 0, maka data semakin murni (homogen).
* **$1$**: Melambangkan total probabilitas maksimal.
* **$\sum$**: Simbol penjumlahan untuk seluruh kelas yang ada.
* **$p_i$**: Probabilitas kemunculan kelas $i$ (Jumlah data kelas $i$ dibagi total data).

### Langkah Perhitungan (Data Keseluruhan):
Berdasarkan gambar data, terdapat 14 baris data dengan hasil: **9 Yes** dan **5 No**.

1.  **Hitung Probabilitas:**
    * $p_{Yes} = 9/14 = 0.642$
    * $p_{No} = 5/14 = 0.357$
2.  **Masukkan ke Rumus:**
    * $Gini(S) = 1 - [(0.642)^2 + (0.357)^2]$
    * $Gini(S) = 1 - [0.412 + 0.127]$
    * $Gini(S) = 1 - 0.539 = \mathbf{0.461}$
