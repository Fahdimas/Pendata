# Data Preparation

Tahap **Data Preparation** adalah langkah krusial di mana data mentah diolah sedemikian rupa agar siap dimasukkan ke dalam model algoritma. Pada tahap ini, kita melakukan pengecekan kualitas data, penanganan nilai yang hilang (missing values), hingga transformasi data jika diperlukan.

## 1. Seleksi Data (Data Selection)
Dalam analisis dataset Iris, semua atribut (`sepal_length`, `sepal_width`, `petal_length`, `petal_width`) dianggap penting karena memiliki korelasi yang signifikan terhadap target `species`. Oleh karena itu, tidak ada kolom yang dihapus pada tahap ini.

## 2. Pembersihan Data (Data Cleaning)
Berdasarkan eksplorasi pada tahap *Data Understanding*, dataset Iris memiliki kualitas yang sangat baik:
* **Missing Values**: Tidak ditemukan data yang kosong (0 missing values).
* **Outliers**: Tidak ditemukan data pencilan yang ekstrim yang dapat mengganggu performa model.
* **Duplikasi**: Pengecekan dilakukan untuk memastikan tidak ada baris data yang terduplikasi secara tidak sengaja.

## 3. Transformasi Data (Data Transformation)
Meskipun data Iris sudah sangat rapi, terkadang kita perlu melakukan normalisasi jika menggunakan algoritma tertentu (seperti KNN). Namun, untuk algoritma berbasis pohon (Tree-based) atau Naive Bayes, data asli seringkali sudah cukup baik untuk digunakan.

## 4. Pelabelan Data (Data Labeling)
Variabel target `species` yang bersifat kategorikal (Setosa, Versicolor, Virginica) akan diubah menjadi format yang dipahami oleh mesin jika diperlukan pada tahap modeling nanti, namun untuk tahap eksplorasi awal, label teks tetap dipertahankan untuk memudahkan interpretasi.
