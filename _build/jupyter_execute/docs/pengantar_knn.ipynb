# Pengantar Algoritma k-Nearest Neighbors (k-NN)

## 1. Konsep Dasar k-NN
**k-Nearest Neighbors (k-NN)** adalah salah satu algoritma *Machine Learning* yang paling fundamental, sederhana, namun sangat efektif. Algoritma ini beroperasi berdasarkan prinsip ketetanggaan: **"Objek yang memiliki karakteristik serupa akan cenderung berada berdekatan satu sama lain dalam ruang multidimensi."**

k-NN memiliki dua sifat utama yang membedakannya dari algoritma lain:
* **Non-Parametric (Non-Parametrik):** k-NN tidak membuat asumsi apa pun tentang distribusi data yang mendasarinya. Algoritma ini sangat berguna di dunia nyata di mana data jarang mengikuti distribusi teoretis yang sempurna (seperti distribusi normal).
* **Lazy Learning (Pembelajaran Malas):** Berbeda dengan algoritma *Eager Learning* (seperti *Decision Tree* atau *Neural Network*) yang membangun model di awal, k-NN tidak memiliki fase pelatihan (*training phase*) yang eksplisit. Algoritma ini hanya menghafal *dataset* latih dan baru melakukan komputasi berat secara *real-time* saat ada data baru yang masuk untuk diuji.

## 2. Fungsi dan Implementasi k-NN
Meskipun k-NN sangat terkenal untuk masalah pengelompokan, fungsinya mencakup tiga ranah utama dalam *Data Science*:
1. **Klasifikasi (Classification):** Menentukan label kategori untuk data baru berdasarkan mayoritas label dari $k$ tetangga terdekatnya.
2. **Regresi (Regression):** Memprediksi nilai numerik kontinu dengan menghitung rata-rata (mean) atau median dari $k$ tetangga terdekatnya.
3. **Imputasi (Missing Value Handling):** Mengisi kekosongan data (*missing data*) dengan mengestimasi nilai berdasarkan karakteristik objek-objek terdekat yang datanya lengkap.

## 3. Metrik Pengukuran Jarak (Distance Metrics) beserta Rumus
Karena k-NN sangat bergantung pada definisi "kedekatan", metrik jarak adalah inti dari algoritma ini. Berikut adalah rumus yang paling umum digunakan beserta penjabaran variabelnya:

### A. Jarak Euclidean (Euclidean Distance)
Ini adalah metrik jarak yang paling umum digunakan untuk data kontinu (numerik). Euclidean menghitung jarak garis lurus terpendek antara dua titik.

**Rumus:**
$$d(x, y) = \sqrt{\sum_{i=1}^{n} (x_i - y_i)^2}$$

**Keterangan Variabel:**
* $d(x,y)$ : Jarak Euclidean antara data target $x$ dan data referensi $y$.
* $n$ : Jumlah total fitur atau atribut (dimensi data) yang digunakan dalam perhitungan.
* $i$ : Indeks fitur ke-$i$ (mulai dari fitur 1 hingga fitur ke-$n$).
* $x_i$ : Nilai fitur ke-$i$ pada data uji / data target (data yang ingin diprediksi).
* $y_i$ : Nilai fitur ke-$i$ pada data latih / data referensi (data tetangga).
* $\sum$ : Simbol Sigma, yang berarti menjumlahkan seluruh hasil perhitungan dari fitur pertama hingga fitur ke-$n$.



### B. Jarak Manhattan (Manhattan Distance)
Sering disebut *City Block Distance*, jarak ini menghitung total selisih absolut antar fitur. Sangat cocok jika data memiliki banyak *outlier* (pencilan) atau berdimensi tinggi.

**Rumus:**
$$d(x, y) = \sum_{i=1}^{n} |x_i - y_i|$$

**Keterangan Variabel Tambahan:**
* $|x_i - y_i|$ : Nilai absolut dari selisih fitur ke-$i$ (nilai negatif akan diubah menjadi positif).

## 4. Menentukan Nilai $k$ (Jumlah Tetangga)
Pemilihan nilai parameter $k$ (jumlah tetangga terdekat yang dievaluasi) sangat menentukan performa model:
* **Jika $k$ terlalu kecil (misal $k=1$):** Model akan sangat sensitif terhadap *noise* atau *outlier* (cenderung *Overfitting*).
* **Jika $k$ terlalu besar:** Model akan terlalu kaku dan mengabaikan pola lokal yang detail (cenderung *Underfitting*), serta memakan waktu komputasi yang jauh lebih lama.

**Aturan Emas (Rule of Thumb) dalam memilih $k$:**
1. Gunakan nilai $k$ ganjil (seperti 3, 5, atau 7) pada kasus klasifikasi biner untuk mencegah terjadinya hasil seri (*tie / draw*) saat proses *voting*.
2. Nilai $k$ yang optimal seringkali didapatkan dari nilai akar kuadrat dari total jumlah sampel data $\sqrt{N}$.

---

## 5. Alur Kerja (Algoritma) k-NN
Secara sistematis, algoritma k-NN mengeksekusi langkah-langkah berikut ketika menerima data baru:
1. **Inisialisasi** nilai $k$ yang diinginkan.
2. **Hitung Jarak** antara data baru tersebut dengan *seluruh* data yang ada di *dataset* referensi menggunakan metrik jarak (misalnya Euclidean).
3. **Urutkan** hasil perhitungan jarak dari nilai terkecil (paling dekat) hingga terbesar (paling jauh).
4. **Saring** dan ambil sejumlah $k$ data teratas yang memiliki jarak terkecil.
5. **Agregasi Hasil**:
   * Jika untuk klasifikasi: Ambil *mode* (nilai yang paling sering muncul).
   * Jika untuk imputasi/regresi: Hitung *mean* (rata-rata) dari nilai tetangga tersebut.
