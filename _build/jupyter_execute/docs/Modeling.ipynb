# Modeling

Tahap **Modeling** adalah inti dari proses data mining. Pada tahap ini, kita menerapkan berbagai algoritma Machine Learning pada data yang telah disiapkan untuk menemukan pola dan membangun model prediksi.

## 1. Pemilihan Algoritma
Dalam proyek klasifikasi Iris ini, beberapa algoritma yang umum digunakan antara lain:
* **K-Nearest Neighbors (KNN)**: Mengklasifikasikan data berdasarkan kemiripan jarak dengan tetangga terdekatnya.
* **Naive Bayes**: Menggunakan pendekatan probabilitas untuk menentukan kelas spesies.
* **Decision Tree**: Membangun model prediksi dalam bentuk struktur pohon keputusan yang mudah diinterpretasikan.

## 2. Skenario Pengujian
Untuk membangun model yang objektif, dataset akan dibagi menjadi dua bagian:
* **Data Training**: Digunakan untuk melatih model agar mengenali pola spesies.
* **Data Testing**: Digunakan untuk menguji sejauh mana model mampu memprediksi data baru yang belum pernah dilihat sebelumnya.
