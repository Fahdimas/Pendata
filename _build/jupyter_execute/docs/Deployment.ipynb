# Deployment

Tahap **Deployment** adalah tahap akhir dari metodologi CRISP-DM. Pada tahap ini, hasil analisis dan model yang telah divalidasi akan diimplementasikan agar dapat memberikan manfaat nyata bagi pengguna atau organisasi.

## 1. Rencana Implementasi
Model klasifikasi Iris yang telah dibangun dapat diimplementasikan dalam berbagai bentuk, seperti:
* **Aplikasi Web**: Membangun antarmuka sederhana di mana pengguna bisa memasukkan ukuran kelopak dan mahkota bunga untuk mendapatkan prediksi spesies secara instan.
* **API Service**: Menyediakan layanan integrasi sehingga sistem lain dapat mengirim data bunga dan menerima hasil klasifikasi secara otomatis.

## 2. Pemeliharaan dan Pemantauan
Setelah model diterapkan, perlu dilakukan pemantauan secara berkala untuk memastikan akurasi tetap terjaga. Jika terdapat perubahan karakteristik data di masa depan, model perlu dilatih ulang (*retraining*) menggunakan data terbaru.

## 3. Laporan Akhir
Seluruh proses dari *Business Understanding* hingga *Evaluation* didokumentasikan dalam website statis ini menggunakan **Jupyter Book**. Dokumentasi ini berfungsi sebagai panduan teknis dan referensi bagi pengembangan proyek di masa mendatang.
